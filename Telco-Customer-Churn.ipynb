```python
from scipy.stats import chi2_contingency, ttest_ind
df = tele_clean.copy()
print(df.shape)   # (7043, 23)
```

```python
#EDA + Statistical Tests
cat_cols = ['gender','SeniorCitizen','Partner','Dependents','PhoneService','MultipleLines',
            'InternetService','OnlineSecurity','OnlineBackup','DeviceProtection','TechSupport',
            'StreamingTV','StreamingMovies','Contract','PaperlessBilling','PaymentMethod','tenure_group']

results = []
for c in cat_cols:
    table = pd.crosstab(df[c], df['Churn'])
    chi2, p, dof, _ = chi2_contingency(table)
    n = table.values.sum()
    cramers_v = np.sqrt(chi2 / (n * (min(table.shape) - 1)))
    results.append([c, round(chi2, 1), p, round(cramers_v, 3)])

chi_df = pd.DataFrame(results, columns=['Feature', 'Chi2', 'p_value', 'Cramers_V'])
chi_df['Significant'] = chi_df['p_value'] < 0.05
chi_df.sort_values('Cramers_V', ascending=False).reset_index(drop=True)
```

```python
top = ['Contract', 'tenure_group', 'InternetService', 'PaymentMethod', 'OnlineSecurity', 'TechSupport']

fig, axes = plt.subplots(2, 3, figsize=(18, 9))
for ax, c in zip(axes.flat, top):
    rate = df.groupby(c)['Churn_flag'].mean().sort_values(ascending=False) * 100
    rate.plot(kind='bar', ax=ax, color='steelblue')
    ax.set_title(f'Churn rate by {c}')
    ax.set_ylabel('Churn %')
    ax.set_xlabel('')
    ax.tick_params(axis='x', rotation=30)
    for i, v in enumerate(rate):
        ax.text(i, v + 1, f'{v:.1f}%', ha='center')
plt.tight_layout()
plt.savefig('churn_by_features.png', dpi=150)
plt.show()
```

## 📊 Phase 4: Key Findings

### 1. Statistical significance (Chi-square + Cramér's V)
| Rank | Feature | Cramér's V | Strength |
|---|---|---|---|
| 1 | Contract | 0.410 | Strong |
| 2 | tenure_group | 0.349 | Moderate |
| 3 | InternetService | 0.322 | Moderate |
| 4 | PaymentMethod | 0.303 | Moderate |
| 5 | PaperlessBilling | 0.191 | Weak |
| 6 | OnlineSecurity | 0.171 | Weak |
| 7 | TechSupport | 0.164 | Weak |

- 15 of 17 categorical features are significantly related to churn (p < 0.05)
- **gender (p = 0.49) and PhoneService (p = 0.34) have no effect on churn**, so
  retention efforts need not target by gender or phone service

### 2. Numeric features (Welch t-test)
| Feature | Churned (mean) | Retained (mean) | p-value |
|---|---|---|---|
| tenure | 18.0 months | 37.6 months | 1.2e-232 |
| MonthlyCharges | $74.4 | $61.3 | 8.6e-73 |
| TotalCharges | $1,531.8 | $2,549.9 | 5.9e-75 |

- Churned customers stayed **less than half as long** and paid **~$13 more per month**
- Profile of a churner: **new customer on a higher monthly bill**

### 3. Correlation
- tenure ↔ Churn: **-0.35** (longer tenure, lower churn)
- MonthlyCharges ↔ Churn: **+0.19** (higher bill, higher churn)
- tenure ↔ TotalCharges: **0.83** → multicollinearity, handle in modelling
  (TotalCharges ≈ tenure × MonthlyCharges)

### 4. Highest-risk segment (Contract × Tenure)
| Contract | 0-12m | 13-24m | 25-48m | 49-72m |
|---|---|---|---|---|
| Month-to-month | **51.4%** | 37.7% | 32.9% | 26.0% |
| One year | 10.5% | 8.1% | 10.6% | 12.9% |
| Two year | 0.0% | 0.0% | 2.2% | 3.3% |

- **Month-to-month customers in their first year churn at 51.4%**, more than 1 in 2
- Even long-term month-to-month customers (49-72m) churn at 26%, nearly
  **8x** two-year customers of the same tenure (3.3%)
- Two-year contract customers almost never churn in their first 2 years (0%)

### 5. Other drivers
- **Internet:** Fiber optic ~42% churn vs DSL ~19%, despite fiber being the premium
  service. Possible pricing or service quality issue
- **Payment:** Electronic check ~45% churn vs ~15-19% for automatic payments and mailed check
- **Add-ons:** No TechSupport ~31% churn vs ~15% with TechSupport

### 6. Business impact
- **Monthly revenue lost to churn: $139,130.85**
- That is **30.5% of total monthly revenue**, about **$1.67M per year**
- Churned customers are 26.5% of the base but 30.5% of revenue, so
  **we are losing higher-paying customers**

### 💡 Recommendations (preliminary)
1. **Convert month-to-month to annual contracts** with a first-year discount,
   targeting customers in their first 12 months
2. **Investigate Fiber optic pricing and service quality**
3. **Push auto-pay** (bank transfer / credit card) over electronic check with a small incentive
4. **Bundle TechSupport / OnlineSecurity** free for the first 3-6 months for new customers

```python
# Preprocessing 
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
```

```python
ids = df['customerID']          # baad me Power BI export ke liye
y   = df['Churn_flag']          # target

X = df.drop(columns=['customerID',   # ID, koi pattern nahi
                     'Churn',        # target ka text version (leakage)
                     'Churn_flag',   # target
                     'tenure_group', # tenure ka hi grouped version (duplicate)
                     'TotalCharges'])# tenure ke saath 0.83 correlation

print("X shape:", X.shape)
print("Columns:", X.columns.tolist())
```

```python
from statsmodels.stats.outliers_influence import variance_inflation_factor

num = df[['tenure', 'MonthlyCharges', 'TotalCharges']]
vif = pd.DataFrame({
    'Feature': num.columns,
    'VIF': [variance_inflation_factor(num.values, i) for i in range(num.shape[1])]
})
vif
```

```python
scale_cols = ['tenure', 'MonthlyCharges']
scaler = StandardScaler()

X_train_sc = X_train.copy()
X_test_sc  = X_test.copy()

X_train_sc[scale_cols] = scaler.fit_transform(X_train[scale_cols])   # fit sirf train pe
X_test_sc[scale_cols]  = scaler.transform(X_test[scale_cols])        # test pe sirf transform

X_train_sc[scale_cols].describe().round(2)
```

## ⚙️ Phase 5: Preprocessing Summary

### 1. Feature Selection
Starting point: 23 columns → kept **18 features** for modelling.

| Dropped Column | Reason |
|---|---|
| customerID | Unique identifier, carries no predictive pattern |
| Churn | Text version of the target; keeping it would cause **data leakage** |
| Churn_flag | This is the target variable (y) |
| tenure_group | Binned version of `tenure`; duplicate information |
| TotalCharges | Approximately tenure × MonthlyCharges; highly correlated with tenure (r = 0.83) |

### 2. Multicollinearity Check (VIF)
| Feature | VIF | Interpretation |
|---|---|---|
| TotalCharges | **8.08** | High (> 5), redundant with tenure and MonthlyCharges |
| tenure | 6.33 | Inflated because of TotalCharges |
| MonthlyCharges | 3.36 | Acceptable |

**Decision:** Dropped `TotalCharges`. It has the highest VIF and its information is already
captured by `tenure` and `MonthlyCharges`. This keeps logistic regression coefficients
stable and interpretable.

### 3. Encoding
- Method: One-hot encoding with `drop_first=True`
- Result: 18 features → **22 model-ready columns** (2 numeric + 20 binary)
- `drop_first=True` avoids the **dummy variable trap** (perfect multicollinearity)
- Baseline (reference) categories: Female, DSL, Month-to-month, Bank transfer (automatic), and "No" for all Yes/No features

### 4. Train-Test Split
| Set | Rows | Columns | Churn % |
|---|---|---|---|
| Train | 5,634 | 22 | 26.54% |
| Test | 1,409 | 22 | 26.54% |

- 80/20 split, `stratify=y` → identical churn ratio in both sets
- `random_state=42` → reproducible results
- customerIDs kept aside (`id_train`, `id_test`) for exporting predictions to Power BI

### 5. Feature Scaling
- `StandardScaler` applied to `tenure` and `MonthlyCharges` only (binary columns already 0/1)
- **Fit on train only**, then applied to test → prevents data leakage
- After scaling (train): mean = 0.00, std = 1.00 for both features
- Two versions maintained:
  - `X_train_sc`, `X_test_sc` → Logistic Regression (scale-sensitive)
  - `X_train`, `X_test` → Random Forest (tree models don't need scaling)

### 6. Class Imbalance Strategy
- Target split: 73.5% No vs 26.5% Yes
- A naive model predicting "No" for everyone would score 73.5% accuracy but catch **zero churners**
- Approach: `class_weight='balanced'` in Phase 6, which penalises missed churners more heavily
- SMOTE not used: the imbalance is moderate, and class weighting avoids synthetic data and leakage risk

### ✅ Ready for Modelling
22 features | 5,634 training rows | 1,409 test rows | No leakage | Balanced evaluation planned

```python
# Modelling 
from sklearn.dummy import DummyClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.pipeline import Pipeline
from sklearn.model_selection import StratifiedKFold, cross_validate
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, roc_auc_score
```

```python
dummy = DummyClassifier(strategy='most_frequent')
dummy.fit(X_train, y_train)
dummy_pred = dummy.predict(X_test)

print("Accuracy :", round(accuracy_score(y_test, dummy_pred), 3))
print("Recall   :", round(recall_score(y_test, dummy_pred), 3))
```

```python
log_reg = LogisticRegression(class_weight='balanced', max_iter=1000, random_state=42)
log_reg.fit(X_train_sc, y_train)

lr_pred  = log_reg.predict(X_test_sc)
lr_proba = log_reg.predict_proba(X_test_sc)[:, 1]
print("Logistic Regression trained ✅")
```

```python
rf = RandomForestClassifier(
    n_estimators=300,
    max_depth=8,
    min_samples_leaf=20,
    class_weight='balanced',
    random_state=42,
    n_jobs=-1
)
rf.fit(X_train, y_train)

rf_pred  = rf.predict(X_test)
rf_proba = rf.predict_proba(X_test)[:, 1]
print("Random Forest trained ✅")
```

```python
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

lr_pipe = Pipeline([
    ('scaler', StandardScaler()),
    ('model', LogisticRegression(class_weight='balanced', max_iter=1000, random_state=42))
])

models_cv = {'Logistic Regression': lr_pipe, 'Random Forest': rf}

for name, model in models_cv.items():
    scores = cross_validate(model, X_train, y_train, cv=cv,
                            scoring=['recall', 'precision', 'roc_auc'])
    print(f"{name:20} | Recall: {scores['test_recall'].mean():.3f} ± {scores['test_recall'].std():.3f}"
          f" | Precision: {scores['test_precision'].mean():.3f}"
          f" | ROC-AUC: {scores['test_roc_auc'].mean():.3f}")
```

```python
def get_metrics(name, y_true, pred, proba=None):
    return {
        'Model': name,
        'Accuracy': accuracy_score(y_true, pred),
        'Precision': precision_score(y_true, pred, zero_division=0),
        'Recall': recall_score(y_true, pred),
        'F1': f1_score(y_true, pred),
        'ROC-AUC': roc_auc_score(y_true, proba) if proba is not None else np.nan
    }

comparison = pd.DataFrame([
    get_metrics('Baseline (Dummy)', y_test, dummy_pred),
    get_metrics('Logistic Regression', y_test, lr_pred, lr_proba),
    get_metrics('Random Forest', y_test, rf_pred, rf_proba)
]).round(3)

comparison
```

### 3. Why do customers churn? Logistic Regression odds ratios
**Increase churn (OR > 1):**
| Feature | Odds Ratio | Meaning |
|---|---|---|
| Fiber optic internet | **3.06** | ~3x the odds of churning vs DSL |

**Reduce churn (OR < 1):**
| Feature | Odds Ratio | Meaning |
|---|---|---|

| One-year contract | 0.486 | 51% lower odds |
| Online Security | 0.710 | 29% lower odds |
| Tech Support | 0.760 | 24% lower odds |

**Note on MonthlyCharges (OR = 0.77):** In EDA, churners had higher bills. In the model, this
effect is absorbed by the services that drive the bill (Fiber, Streaming). After controlling for
them, the remaining effect of the bill alone is slightly negative. This is **confounding**, not
a contradiction.

**Gender (OR = 1.04)** has almost no effect, consistent with the chi-square test (p = 0.49).
